```python
def prediction_heatmap(sermon: Sermon, csv_list: list) -> list:
    sentences = []
    color_map = {'lieder':[192, 54, 157], 'bibel':[246, 169, 122]}
    for i in range(len(sermon.chunked)):
        for j in range(len(sermon.chunked[i])):
            bibel = False
            lied = False
            for df, task in csv_list:
                row = df[(df['Paragraph'] == i) & (df['Satz'] == j)]
                if not row.empty:
                    row_dict = row.iloc[0].to_dict()
                    if task == "bibel":
                        bibel = True
                        #bibel = row_dict['Ähnlichkeit']
                    else:
                        lied = True
                        #lied = row_dict['Ähnlichkeit']
        
        if bibel and lied:
            elem = '<div class="sermon_distant" style="background-image: linear-gradient(to right,rgba(246, 169, 122) 0%, rgba(192, 54, 157) 100%)"></div>'
        elif bibel:
            elem = '<div class="sermon_distant" style="background-color: rgb(246, 169, 122)"></div>'
        elif lied:
            elem = '<div class="sermon_distant" style="background-color: rgb(192, 54, 157)"></div>'
        else:
            elem = '<div class="sermon_distant" style="background-color: LightGray;"></div>'
        sentences.append(elem)

    return sentences
```
